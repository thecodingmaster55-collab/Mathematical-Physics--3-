```python
from collections import deque

def bfs(start, goal):
    queue = deque([(start, [])])
    visited = {tuple(start)}

    moves = [
        (-1, 0),  # Up
        (1, 0),   # Down
        (0, -1),  # Left
        (0, 1)    # Right
    ]

    while queue:
        state, path = queue.popleft()

        if state == goal:
            return path + [state]

        zero = state.index(0)
        row, col = divmod(zero, 3)

        for dr, dc in moves:
            nr, nc = row + dr, col + dc

            if 0 <= nr < 3 and 0 <= nc < 3:
                new_zero = nr * 3 + nc
                new_state = state[:]

                new_state[zero], new_state[new_zero] = (
                    new_state[new_zero], new_state[zero]
                )

                key = tuple(new_state)

                if key not in visited:
                    visited.add(key)
                    queue.append((new_state, path + [state]))

    return None


start = [1, 2, 3,
         0,4 ,5,
         6, 7, 8]

goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]

result = bfs(start, goal)

if result:
    print("Solution found:")
    for state in result:
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()
    print("Number of moves:", len(result) - 1)
else:
    print("No solution exists.")
```
