```python
import cmath


def get_nth_roots(z, n):
  r, theta = cmath.polar(z)
  r_nth = r ** (1 / n)
  roots = []

  for k in range(n):
    angle = (theta + 2 * cmath.pi * k) / n

    polar_root = (r_nth, angle)
    cartesian_root = cmath.rect(r_nth, angle)

    roots.append((cartesian_root, polar_root))

  return roots

z = 8 + 0j
n = 3
results = get_nth_roots(z, n)

for i, (cart, pol) in enumerate(results):
  r_val, theta_val = pol
  print(f"Root {i+1}:")
  print(f"  Cartesian : {cart:.4f}")
  print(f"  Polar     : r = {r_val:.4f}, theta = {theta_val:.4f} rad")
```

```python
import cmath
import math


z = 3 + 4j
print(f"Original Vector: {z} -> (Coordinates: {z.real}, {z.imag})")

translation_vector = 2 + 1j
z_translated = z + translation_vector
print(f"Translated:      {z_translated}")

scale_factor = 2.0
z_scaled = z * scale_factor
print(f"Scaled (x2):     {z_scaled}")


angle = math.pi / 2
rotation_multiplier = cmath.exp(1j * angle)
z_rotated = z * rotation_multiplier
print(f"Rotated 90°:     {z_rotated.real:.1f} + {z_rotated.imag:.1f}j")


z_reflected_x = z.conjugate()
print(f"Reflected (X):   {z_reflected_x}")

z_reflected_y = -z.conjugate()
print(f"Reflected (Y):   {z_reflected_y}")
```
