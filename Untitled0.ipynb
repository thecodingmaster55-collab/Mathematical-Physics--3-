```python
import math
import matplotlib.pyplot as plt


x = 2
y = 3


x1 = x + 2
y1 = y + 1


x2 = x * 2
y2 = y * 2


theta = math.radians(90)

x3 = x * math.cos(theta) - y * math.sin(theta)
y3 = x * math.sin(theta) + y * math.cos(theta)


x4 = x
y4 = -y


print("Original point:", (x, y))
print("Translation:", (x1, y1))
print("Scaling:", (x2, y2))
print("Rotation:", (round(x3, 2), round(y3, 2)))
print("Reflection:", (x4, y4))


points = [(x, y), (x1, y1), (x2, y2),
          (x3, y3), (x4, y4)]

labels = ["Original", "Translation", "Scaling",
          "Rotation", "Reflection"]

for p, label in zip(points, labels):
    plt.scatter(p[0], p[1], label=label)

plt.xlabel("Real Axis (X)")
plt.ylabel("Imaginary Axis (Y)")
plt.title("2-D Vector Transformations")

plt.axhline(0)
plt.axvline(0)

plt.grid()
plt.legend()
plt.show()
```
