```python
import numpy as np
import matplotlib.pyplot as plt
a= 1000
b=500
n = 500
h = (a-b)/n
X= np.zeros (n)
A = np.zeros((n,n))
R = np.zeros(n)
x0 = 0
p = 0
q=0
for i in range (1,n+1):
  X[i-1]=x0+i*h
  A[i-1,i-1]= -2+q*(h**2)
  R[i-1]= 5*(h**2)
  if i>1:
    A[i-1,i-2]=1-(h*p/2)
  if i!=n:
    A[i-1,i]=1+(h*p/2)

print(A)
B = np.linalg.solve(A,R)
print(B)
print(X)
plt.plot(X,R)
plt.grid()
plt.show()
```
