```python
import cmath


def get_nth_roots(z, n):
  r, theta = cmath.polar(z)
  r_nth = r ** (1 / n)
  roots = []

  for k in range(n):
    angle = (theta + 2 * cmath.pi * k) / n

    polar_root = (r_nth, angle)
    cartesian_root = cmath.rect(r_nth, angle)

    roots.append((cartesian_root, polar_root))

  return roots

z = 8 + 0j
n = 3
results = get_nth_roots(z, n)

for i, (cart, pol) in enumerate(results):
  r_val, theta_val = pol
  print(f"Root {i+1}:")
  print(f"  Cartesian : {cart:.4f}")
  print(f"  Polar     : r = {r_val:.4f}, theta = {theta_val:.4f} rad")
```
